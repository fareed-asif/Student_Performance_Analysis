```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

cleaned = pd.read_csv("/content/cleaned_students.csv")
score_cols = ['math score', 'reading score', 'writing score']

sns.set_style("whitegrid")
plt.rcParams['figure.dpi'] = 100
print("Ready to plot!")
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(
    cleaned['math score'],
    kde=True,
    color='steelblue',
    bins=20,
    edgecolor='white'
)
plt.title("Distribution of Math Scores", fontsize=14, fontweight='bold')
plt.xlabel("Math Score", fontsize=12)
plt.ylabel("Number of Students", fontsize=12)
plt.grid(True, alpha=0.5)
plt.tight_layout()
plt.savefig("/content/plot1_histogram.png", dpi=150)
plt.show()
print("Plot 1 saved!")
```

```python
plt.figure(figsize=(7, 5))
sns.boxplot(
    x='gender',
    y='math score',
    data=cleaned,
    palette='Set2',
    width=0.5
)
plt.title("Math Score Distribution by Gender", fontsize=14, fontweight='bold')
plt.xlabel("Gender", fontsize=12)
plt.ylabel("Math Score", fontsize=12)
plt.grid(True, alpha=0.5, axis='y')
plt.tight_layout()
plt.savefig("/content/plot2_boxplot.png", dpi=150)
plt.show()
print("Plot 2 saved!")
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(
    x='reading score',
    y='writing score',
    hue='gender',
    data=cleaned,
    alpha=0.6,
    palette={'male': 'steelblue', 'female': 'coral'},
    s=60
)
plt.title("Reading Score vs Writing Score by Gender", fontsize=14, fontweight='bold')
plt.xlabel("Reading Score", fontsize=12)
plt.ylabel("Writing Score", fontsize=12)
plt.grid(True, alpha=0.5)
plt.legend(title='Gender')
plt.tight_layout()
plt.savefig("/content/plot3_scatter.png", dpi=150)
plt.show()
print("Plot 3 saved!")
```

```python
avg = cleaned.groupby('parental level of education')['math score'].mean().sort_values()

plt.figure(figsize=(10, 5))
bars = avg.plot(
    kind='bar',
    color='coral',
    edgecolor='black',
    width=0.6
)
plt.title("Average Math Score by Parental Education Level", fontsize=14, fontweight='bold')
plt.xlabel("Parental Education", fontsize=12)
plt.ylabel("Average Math Score", fontsize=12)
plt.xticks(rotation=30, ha='right')
plt.grid(True, alpha=0.5, axis='y')
plt.ylim(0, 100)
plt.tight_layout()
plt.savefig("/content/plot4_bar.png", dpi=150)
plt.show()
print("Plot 4 saved!")
```

```python
avg = cleaned.groupby('parental level of education')['math score'].mean().sort_values()

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle(
    "Student Performance Analysis Dashboard",
    fontsize=18, fontweight='bold', y=1.01
)

# Top-left: Histogram
sns.histplot(cleaned['math score'], kde=True, ax=axes[0,0], color='steelblue', bins=20)
axes[0,0].set_title("Math Score Distribution", fontweight='bold')
axes[0,0].set_xlabel("Math Score")
axes[0,0].set_ylabel("Count")
axes[0,0].grid(True, alpha=0.5)

# Top-right: Boxplot
sns.boxplot(x='gender', y='math score', data=cleaned, ax=axes[0,1], palette='Set2')
axes[0,1].set_title("Math Score by Gender", fontweight='bold')
axes[0,1].set_xlabel("Gender")
axes[0,1].set_ylabel("Math Score")
axes[0,1].grid(True, alpha=0.5, axis='y')

# Bottom-left: Scatter
sns.scatterplot(x='reading score', y='writing score', hue='gender',
    data=cleaned, ax=axes[1,0], alpha=0.6,
    palette={'male':'steelblue','female':'coral'})
axes[1,0].set_title("Reading vs Writing Score", fontweight='bold')
axes[1,0].set_xlabel("Reading Score")
axes[1,0].set_ylabel("Writing Score")
axes[1,0].grid(True, alpha=0.5)

# Bottom-right: Bar
avg.plot(kind='bar', ax=axes[1,1], color='coral', edgecolor='black')
axes[1,1].set_title("Avg Math by Parental Education", fontweight='bold')
axes[1,1].set_xlabel("Parental Education")
axes[1,1].set_ylabel("Avg Math Score")
axes[1,1].set_xticklabels(avg.index, rotation=30, ha='right', fontsize=9)
axes[1,1].grid(True, alpha=0.5, axis='y')

plt.tight_layout()
plt.savefig("/content/best_visualization.png", dpi=150, bbox_inches='tight')
plt.show()
print("best_visualization.png saved!")
```
