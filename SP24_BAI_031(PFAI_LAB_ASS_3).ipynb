```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("/content/StudentsPerformance.csv")

print("Shape (rows, columns):", df.shape)
print("\nFirst 5 rows:")
print(df.head())
print("\nColumn names:", list(df.columns))
print("\nData types:")
print(df.dtypes)
```

```python
score_cols = ['math score', 'reading score', 'writing score']

print("=== MEAN ===")
print(df[score_cols].mean().round(2))

print("\n=== MEDIAN ===")
print(df[score_cols].median())

print("\n=== MODE ===")
print(df[score_cols].mode().iloc[0])

print("\n=== STANDARD DEVIATION ===")
print(df[score_cols].std().round(2))

print("\n=== MIN ===")
print(df[score_cols].min())

print("\n=== MAX ===")
print(df[score_cols].max())

print("\n=== FULL SUMMARY ===")
print(df[score_cols].describe().round(2))
```

```python
print("=== Mean vs Median Comparison ===")
print(f"{'Subject':<20} {'Mean':>8} {'Median':>8} {'Difference':>12}")
print("-" * 50)

for col in score_cols:
    mean = df[col].mean()
    median = df[col].median()
    diff = mean - median
    print(f"{col:<20} {mean:>8.2f} {median:>8.2f} {diff:>12.2f}")
```

```python
print("=== Average Scores by Gender ===")
gender_stats = df.groupby('gender')[score_cols].mean().round(2)
print(gender_stats)

print("\n=== Count by Gender ===")
print(df['gender'].value_counts())
```

```python
print("=== Average Scores by Parental Education ===")
parent_stats = df.groupby('parental level of education')[score_cols].mean().round(2)
print(parent_stats)

print("\n=== Count by Parental Education ===")
print(df['parental level of education'].value_counts())
```

1. Female students score higher in reading (72.6) and writing (72.5) compared to males, while males score slightly higher in math (69.0 vs 63.6).

2. Students whose parents have a master's degree score the highest across all subjects (math: 69.7, reading: 75.4, writing: 75.7).

3. Mean and Median are very close for all three scores (difference < 1), indicating the data is approximately symmetric with no extreme skew.

4. Reading and writing scores are highly correlated — students who score high in one tend to score high in the other.

5. Standard deviation (~15 points) across all subjects shows significant variation in student performance.
